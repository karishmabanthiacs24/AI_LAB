```python


board = ['1', '2', '3',
         '4', '5', '6',
         '7', '8', '9']


def display_board():
    print()
    print(" ", board[0], "|", board[1], "|", board[2])
    print("---|---|---")
    print(" ", board[3], "|", board[4], "|", board[5])
    print("---|---|---")
    print(" ", board[6], "|", board[7], "|", board[8])
    print()


def check_winner():

    if board[0] == board[1] == board[2]:
        return True
    if board[3] == board[4] == board[5]:
        return True
    if board[6] == board[7] == board[8]:
        return True

    if board[0] == board[3] == board[6]:
        return True
    if board[1] == board[4] == board[7]:
        return True
    if board[2] == board[5] == board[8]:
        return True


    if board[0] == board[4] == board[8]:
        return True
    if board[2] == board[4] == board[6]:
        return True

    return False


def game():
    player = 'X'
    moves = 0

    print("===== TIC-TAC-TOE =====")

    while True:
        display_board()

        print("Player", player, "turn")

        try:
            position = int(input("Enter position (1-9): "))
        except ValueError:
            print("Please enter a number!")
            continue

        if position < 1 or position > 9:
            print("Invalid position! Choose between 1 and 9.")
            continue

        index = position - 1

        if board[index] == 'X' or board[index] == 'O':
            print("Position already occupied!")
            continue

        board[index] = player
        moves += 1

        if check_winner():
            display_board()
            print("Player", player, "wins!")
            break

        if moves == 9:
            display_board()
            print("Game Draw!")
            break

        if player == 'X':
            player = 'O'
        else:
            player = 'X'


game()
```

```python
# Vacuum Cleaner Agent

roomA = input("Enter Room A status (Dirty/Clean): ")
roomB = input("Enter Room B status (Dirty/Clean): ")

location = input("Enter vacuum location (A/B): ").upper()

print("\n--- Vacuum Cleaner Agent ---")

# Simple Reflex Agent
while roomA == "Dirty" or roomB == "Dirty":

    if location == "A":

        if roomA == "Dirty":
            print("Room A is Dirty -> Suck")
            roomA = "Clean"
        else:
            print("Room A is Clean -> Move Right")
            location = "B"

    else:

        if roomB == "Dirty":
            print("Room B is Dirty -> Suck")
            roomB = "Clean"
        else:
            print("Room B is Clean -> Move Left")
            location = "A"

print("\nBoth rooms are Clean!")
print("Goal Achieved!")
```
