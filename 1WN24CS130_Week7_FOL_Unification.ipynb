```python
def is_variable(x):
    return x.islower()


def unify(x, y, substitution):

    if x == y:
        return substitution

    if is_variable(x):
        substitution[x] = y
        return substitution

    if is_variable(y):
        substitution[y] = x
        return substitution

    return None


def unification():

    print("UNIFICATION IN FIRST ORDER LOGIC")
    print("--------------------------------")

    e1 = input("Enter first expression: ")
    e2 = input("Enter second expression: ")

    # Separate predicate and arguments
    p1 = e1[:e1.index("(")]
    p2 = e2[:e2.index("(")]

    a1 = e1[e1.index("(")+1:e1.index(")")].split(",")
    a2 = e2[e2.index("(")+1:e2.index(")")].split(",")

    # Check predicates
    if p1 != p2:
        print("Unification Failed: Different predicates")
        return

    # Check number of arguments
    if len(a1) != len(a2):
        print("Unification Failed: Different number of arguments")
        return

    substitution = {}

    # Compare arguments
    for x, y in zip(a1, a2):

        x = x.strip()
        y = y.strip()

        result = unify(x, y, substitution)

        if result is None:
            print("Unification Failed")
            return

    print("\nUnification Successful!")
    print("Substitution:", substitution)


unification()








```
